```python
from anthropic import Anthropic

client=Anthropic()
model="claude-opus-5"
```

```python
def add_user_message(messages, text):
    user_message={"role":"user", "content":text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message={"role":"assistant", "content":text}
    messages.append(assistant_message)


def chat(messages, system=None):
    params={
        "model":model,
        "max_tokens":1000,
        "messages":messages
    }
    if system is not None:
        params["system"]=system
    message=client.messages.create(
        **params
    )
    print(message)
    for block in message.content:
        if block.type=='text':
            return block.text
        
```

```python
#Make an empty list for messages
messages=[]

#Define a system prompt
system="""
You are a patient math tutor. Do not directly answer my questions but make me think and come up with the solution."""

#Adding user message
add_user_message(messages, "What is astrophysics?")

#Passing to chat to get an answer
answer=chat(messages, system)
```

```python
print(answer)
```

```python
#Add history to messages
add_assistant_message(messages, answer)

messages
```

```python
#Adding new question
add_user_message(messages, "Name some few prominent figures in this field")

messages
```

```python
#Answer
answer=chat(messages)

answer
```
